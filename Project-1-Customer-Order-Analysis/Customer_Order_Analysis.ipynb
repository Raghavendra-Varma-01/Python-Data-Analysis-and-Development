# Analyzing Customer Orders Using Python

## Course-End Project

### Project Objective

This project analyzes customer orders for an e-commerce company using Pythons built-in data structures and control structures.

The analysis focuses on:

- Storing and organizing customer order data
- Classifying products into different categories
- Calculating total spending for each customer
- Classifying customers based on their spending
- Calculating revenue for each product category
- Identifying unique products and product categories
- Finding customers who purchased electronics
- Identifying the top three highest-spending customers
- Analyzing customers who purchased products from multiple categories
- Generating useful business insights from customer purchasing patterns

### Python Concepts Used

This project demonstrates the use of:

- Lists
- Tuples
- Dictionaries
- Sets
- Loops
- Conditional statements
- List comprehensions
- Sorting
                      

```python
# Customer names
customers = [
    "Alice",
    "Bob",
    "Charlie",
    "David",
    "Emma",
    "Frank",
    "Grace",
    "Henry",
    "Isha"
]

# Customer order data
# Each order is stored as:
# (Customer Name, Product, Price, Category)

orders = [
    ("Alice", "Laptop", 800, "Electronics"),
    ("Alice", "Headphones", 50, "Electronics"),
    ("Alice", "T-Shirt", 30, "Clothing"),

    ("Bob", "Smartphone", 600, "Electronics"),
    ("Bob", "Jeans", 60, "Clothing"),

    ("Charlie", "Microwave", 120, "Home Essentials"),
    ("Charlie", "T-Shirt", 25, "Clothing"),

    ("David", "Laptop", 800, "Electronics"),
    ("David", "Smartwatch", 150, "Electronics"),
    ("David", "Jacket", 90, "Clothing"),

    ("Emma", "Sofa", 450, "Home Essentials"),
    ("Emma", "Bedsheet", 40, "Home Essentials"),

    ("Frank", "Headphones", 50, "Electronics"),
    ("Frank", "T-Shirt", 25, "Clothing"),

    ("Grace", "Refrigerator", 700, "Home Essentials"),
    ("Grace", "Jeans", 60, "Clothing"),

    ("Henry", "Keyboard", 70, "Electronics"),
    ("Henry", "Mouse", 30, "Electronics"),
    ("Henry", "T-Shirt", 25, "Clothing"),

    ("Isha", "T-Shirt", 25, "Clothing")
]
```

```python
# Create a dictionary to store products ordered by each customer

customer_orders = {}

for customer, product, price, category in orders:
    if customer not in customer_orders:
        customer_orders[customer] = []
    
    customer_orders[customer].append(product)

print("Customer Orders Dictionary:")
for customer, products in customer_orders.items():
    print(f"{customer}: {products}")
```

```python
# Create a dictionary mapping each product to its category

product_category = {
    "Laptop": "Electronics",
    "Headphones": "Electronics",
    "Smartphone": "Electronics",
    "Smartwatch": "Electronics",
    "Keyboard": "Electronics",
    "Mouse": "Electronics",
    "T-Shirt": "Clothing",
    "Jeans": "Clothing",
    "Jacket": "Clothing",
    "Microwave": "Home Essentials",
    "Sofa": "Home Essentials",
    "Bedsheet": "Home Essentials",
    "Refrigerator": "Home Essentials"
}

# Create a set containing unique product categories
unique_categories = set(product_category.values())

print("Product Categories:")
for product, category in product_category.items():
    print(f"{product}: {category}")

print("\nUnique Product Categories:")
print(unique_categories)
```

```python
# Calculate total spending for each customer

customer_spending = {}

for customer, product, price, category in orders:
    if customer not in customer_spending:
        customer_spending[customer] = 0
    
    customer_spending[customer] += price


# Classify customers based on their total spending

customer_classification = {}

for customer, total in customer_spending.items():
    if total > 100:
        customer_classification[customer] = "High-value"
    elif total >= 50:
        customer_classification[customer] = "Moderate"
    else:
        customer_classification[customer] = "Low-value"


# Display customer spending and classification

print("Customer Spending and Classification")
print("-" * 45)

for customer in customers:
    print(
        f"{customer}: ${customer_spending[customer]:.2f} "
        f"- {customer_classification[customer]}"
    )
```

```python
# Calculate total revenue for each product category

category_revenue = {}

for customer, product, price, category in orders:
    if category not in category_revenue:
        category_revenue[category] = 0
    
    category_revenue[category] += price


# Display category-wise revenue

print("Revenue by Product Category")
print("-" * 35)

for category, revenue in category_revenue.items():
    print(f"{category}: ${revenue:.2f}")
```

```python
# Extract unique products from all customer orders

unique_products = set()

for customer, product, price, category in orders:
    unique_products.add(product)

print("Unique Products:")
for product in sorted(unique_products):
    print(product)

print("\nTotal Number of Unique Products:", len(unique_products))
```

```python
# Find customers who purchased electronics

electronics_customers = [
    customer
    for customer in customers
    if any(category == "Electronics" for cust, product, price, category in orders if cust == customer)
]

print("Customers who purchased Electronics:")
for customer in electronics_customers:
    print(customer)
```

```python
# Find the top three highest-spending customers

top_customers = sorted(
    customer_spending.items(),
    key=lambda item: item[1],
    reverse=True
)

print("Top 3 Highest-Spending Customers:")
print("-" * 40)

for rank, (customer, total) in enumerate(top_customers[:3], start=1):
    print(f"{rank}. {customer}: ${total:.2f}")
```

```python
# Create sets of customers for each product category

electronics_customers_set = set()
clothing_customers_set = set()
home_essentials_customers_set = set()

for customer, product, price, category in orders:
    if category == "Electronics":
        electronics_customers_set.add(customer)
    elif category == "Clothing":
        clothing_customers_set.add(customer)
    elif category == "Home Essentials":
        home_essentials_customers_set.add(customer)


print("Electronics Customers:")
print(electronics_customers_set)

print("\nClothing Customers:")
print(clothing_customers_set)

print("\nHome Essentials Customers:")
print(home_essentials_customers_set)
```

```python
# Find customers who purchased both Electronics and Clothing

electronics_and_clothing = (
    electronics_customers_set & clothing_customers_set
)

print("Customers who purchased both Electronics and Clothing:")
print(electronics_and_clothing)
```

```python
# Find customers who purchased products from multiple categories

customer_categories = {}

for customer, product, price, category in orders:
    if customer not in customer_categories:
        customer_categories[customer] = set()
    
    customer_categories[customer].add(category)


print("Customers who purchased from multiple categories:")
print("-" * 50)

for customer, categories in customer_categories.items():
    if len(categories) > 1:
        print(f"{customer}: {categories}")
```

```python
# Count how many times each product was purchased

product_frequency = {}

for customer, product, price, category in orders:
    if product not in product_frequency:
        product_frequency[product] = 0
    
    product_frequency[product] += 1


# Sort products by purchase frequency

most_purchased_products = sorted(
    product_frequency.items(),
    key=lambda item: item[1],
    reverse=True
)

print("Product Purchase Frequency")
print("-" * 35)

for product, count in most_purchased_products:
    print(f"{product}: {count} purchase(s)")
```

```python
# Display a final summary of customer spending and classification

print("Customer Purchase Summary")
print("=" * 50)

for customer in customers:
    total = customer_spending[customer]
    classification = customer_classification[customer]
    
    print(f"Customer: {customer}")
    print(f"Total Spending: ${total:.2f}")
    print(f"Classification: {classification}")
    print("-" * 50)
```

```python
# Generate a final numerical analysis of the customer orders

# Count customers in each spending category
high_value_count = 0
moderate_count = 0
low_value_count = 0

for customer, classification in customer_classification.items():
    if classification == "High-value":
        high_value_count += 1
    elif classification == "Moderate":
        moderate_count += 1
    else:
        low_value_count += 1


# Find the highest-revenue category
highest_revenue_category = max(
    category_revenue,
    key=category_revenue.get
)

# Find the most frequently purchased product
most_frequent_product = max(
    product_frequency,
    key=product_frequency.get
)


# Display final analysis
print("FINAL BUSINESS ANALYSIS")
print("=" * 50)

print("\nCustomer Classification:")
print(f"High-value customers: {high_value_count}")
print(f"Moderate customers: {moderate_count}")
print(f"Low-value customers: {low_value_count}")

print("\nCategory Revenue:")
for category, revenue in category_revenue.items():
    print(f"{category}: ${revenue:.2f}")

print(f"\nHighest Revenue Category: {highest_revenue_category}")
print(
    f"Highest Revenue: "
    f"${category_revenue[highest_revenue_category]:.2f}"
)

print(f"\nMost Frequently Purchased Product: {most_frequent_product}")
print(
    f"Number of Purchases: "
    f"{product_frequency[most_frequent_product]}"
)

print("\nTop 3 Highest-Spending Customers:")
for rank, (customer, total) in enumerate(top_customers[:3], start=1):
    print(f"{rank}. {customer}: ${total:.2f}")

print("\nCustomers Who Bought Both Electronics and Clothing:")
print(electronics_and_clothing)
```

## Business Insights

Based on the analysis of customer orders, the following business insights were identified:

### 1. Customer Classification

The analysis identified 7 high-value customers, 1 moderate customer, and 1 low-value customer. This shows that most customers in the dataset have relatively high spending levels.

High-value customers can be targeted with loyalty programs, personalized offers, and exclusive promotions to encourage continued purchases.

### 2. Category-wise Revenue

Electronics generated the highest revenue with total sales of **$2,550.00**.

Home Essentials generated **$1,310.00**, while Clothing generated **$340.00**.

This indicates that Electronics is the strongest-performing category in terms of revenue. The company should ensure sufficient inventory of electronic products and consider targeted marketing campaigns for this category.

### 3. Most Frequently Purchased Product

The **T-Shirt** was the most frequently purchased product, with **5 purchases**.

This suggests consistent demand for T-Shirts and indicates that maintaining sufficient stock could help avoid inventory shortages.

### 4. Top Spending Customers

The three highest-spending customers were:

1. **David** — $1,040.00
2. **Alice** — $880.00
3. **Grace** — $760.00

These customers can be considered important high-value customers and could be targeted with personalized marketing and loyalty rewards.

### 5. Cross-Category Purchasing Behavior

Five customers — **Alice, Bob, David, Frank, and Henry** — purchased products from both Electronics and Clothing.

This indicates an opportunity for cross-selling. For example, customers purchasing electronic products could be offered relevant clothing promotions or bundled offers.

### 6. Overall Business Recommendation

The business should focus on retaining high-value customers, maintaining adequate inventory for frequently purchased products, and using cross-category purchasing patterns to create targeted marketing campaigns.

The category-wise revenue analysis also suggests that Electronics should receive particular attention because it generated the highest revenue in the dataset.
